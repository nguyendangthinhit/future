```python
def arithmetic_sequence(a1,d,n):
    return a1 + (n-1)*d 
x=[]
for i in range(1,6):
    x.append(arithmetic_sequence(1,3,i))
print(x)
print(sum(x))
```

```python
def geometric_sequence(a1,r,n):
    return a1 * r**(n-1)
x=[]
for i in range(1,6):
    x.append(geometric_sequence(1,2,i))
print(x)
print(sum(x))
```

```python
def my_abs(x):
    if x >= 0:
        return x
    else:
        return -x
print(my_abs(5))
print(my_abs(-5))
```

BAI 6

```python
from math import sqrt
def euclidean_distance(A,B):
    x1,y1 = A
    x2,y2 = B
    return round(sqrt((x1-x2)**2+(y1-y2)**2), 2)
print(euclidean_distance((0,0),(3,4)))

```

BAI 7

```python
a=[1,2,3,4,5]
b=[4,5,6,7]
def set_oparation(a,b):
    giao = set(a) & set(b)
    hop = set(a) | set(b)
    hieuab = set(a) - set(b)
    hieuba = set(b) - set(a)
    return giao,hop,hieuab,hieuba
    
giao,hop,hieuab,hieuba = set_oparation(a,b)   
print("Giao:", giao)
print("Hợp:", hop)
print("Hiệu A-B:", hieuab)
print("Hiệu B-A:", hieuba)
```

```python
angles=[0, 30, 45, 60, 90]
def trig_table(angles):
    for angle in angles:
        print(f"sin({angle}) = {math.sin(math.radians(angle))}")
        print(f"cos({angle}) = {math.cos(math.radians(angle))}")
trig_table(angles)
```
